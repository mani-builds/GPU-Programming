```python
import torch
torch.__version__
import triton
import triton.language as tl
```

```python
M, N = 1024, 2048
```

```python
a = torch.rand((M, N), device='cuda', dtype=torch.float32)
```

```python
a.is_contiguous()
print(a.is_contiguous())
print(a.stride(0), a.stride(1))
print(a.storage_offset())
```

```python
a_t = a.t()
print(a_t.is_contiguous())
print(a_t.stride(0), a_t.stride(1))
print(a_t.storage_offset())
```

```python
a_t_ctgs = a.t().contiguous()
print(a_t_ctgs.is_contiguous())
print(a_t_ctgs.stride(0), a_t_ctgs.stride(1))
print(a_t_ctgs.storage_offset())
```
