```python
def tiledMatmul(a, b, c, M, N, K, TILE):
    for row_out in range(0, M, TILE):
        for col_out in range(0, N, TILE):
            for k_out in range(0, K, TILE):
                # computer the boundaries of current tile
                row_max = min(row_out + TILE, M)
                col_max = min(col_out + TILE, N)
                k_max = min(k_out + TILE, K)
                # multiply tiles
                for i in range(row_out, row_max):
                    for j in range(col_out, col_max):
                        sum = 0
                        for k in range(k_out, k_max):
                            sum += a[i][k] * b[k][j]
                        c[i][j] += sum
    return c
```

```python

def matmul(a, b, c, M, N, K):
    for row_out in range(0, M):
        for col_out in range(0, N):
            # sum = 0
            # for i in range(0, K):
            #     sum += a[row_out][i] * b[i][col_out]
            # c[row_out][col_out] = sum
            c[row_out][col_out] = np.dot(a[row_out,:] , b[:,col_out])
    return c
```

```python
def naive_softmax(x):
    y = torch.empty_like(x)
    x_max = torch.max(x)
    num = torch.exp(x - x_max)
    den = torch.sum(num)
    y = num / den
    return y
```

```python

def top_k_masking(logits, k):
    values, _ = torch.topk(logits, k)

    cutoff_vals = values[:, -1].unsqueeze(-1)

    mask = logits < cutoff_vals
    logits[mask] = float('-inf')
    return logits
```

```python
# Example usage:
# Batch size 2, Vocab size 8, Top-K = 3
logits = torch.randn(2, 8)
filtered_logits = top_k_masking(logits, k=3)
filtered_logits
```

```python
def layer_norm(x, w, b):
    M, N = x.shape
    y = torch.empty_like(x)
    x_mean = torch.sum(x)
    num = x - x_mean
    var = num * num
    var = torch.sum(var)
    rstd = 1 / torch.sqrt(var + torch.e)
    y = num * rstd * w + b
    return y
```
