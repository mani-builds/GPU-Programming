```python
import torch
torch.__version__
import torch.nn as nn
from torch.utils.checkpoint import checkpoint
```

```python
class MemoryEfficientResNet(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.conv_block = nn.Sequential(
            nn.Conv2d(dim, dim, padding=1, bias=False),
            nn.BatchNorm2d(dim),
            nn.ReLU(),
            nn.Conv2d(dim, dim, padding=1, bias=False),
            nn.BatchNorm2d(dim)
        )
    def forward(self, x):
        # We wrap the heavy computation in the checkpoint function
        return x + checkpoint(self.conv_block, x, use_reentrant=False)
```

```python
# usage
model = nn.Sequential(*[MemoryEfficientResNet(64) for _ in range(100)])
```
