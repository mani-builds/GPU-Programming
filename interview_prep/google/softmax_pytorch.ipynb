```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(12)
x = torch.rand(10, device='cuda')
```

```python
def naive_softmax(x):
    y = torch.empty_like(x)
    x_max = torch.max(x)
    num = torch.exp(x - x_max)
    den = torch.sum(num)
    y = num / den
    return y
```

```python
naive_softmax(x)
```

```python
torch.softmax(x, dim=0)
```
